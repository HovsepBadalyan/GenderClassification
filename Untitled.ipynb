```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
import os
import cv2
import matplotlib.image as mpimg
```

```python
label = {
    'men': 0,
    'women': 1
}
```

```python
X_train = []
y_train = []
for i in os.listdir('gender_dataset/'):
        if i not in label:
            continue
        for j in os.listdir(f'gender_dataset/{i}/'):
            
            
            
            image = mpimg.imread(f'gender_dataset/{i}/{j}')
            image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
            image = cv2.resize(image, (200, 200))
            image = image/255
            image = image.reshape(200,200)
            X_train.append(image)
            y_train.append(label[i])
            
X_train = np.array(X_train)
y_train = np.array(y_train)
```

```python
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

```

```python
from sklearn.utils import shuffle
```

```python
X_train, y_train = shuffle(X_train, y_train)
```

```python
from keras.utils import to_categorical
```

```python
y_train = to_categorical(y_train, num_classes=2)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split( X_train, y_train, test_size=0.2, random_state=42)
```

```python
from keras import Sequential, Input
from keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
from keras.optimizers import Adam
from keras import regularizers
```

```python
model = Sequential()
model.add(Input(shape=(200, 200, 1)))
model.add(Conv2D(filters=32, kernel_size=(3, 3), activation='relu', padding='same'))
model.add(MaxPooling2D(pool_size=(2, 2)))
model.add(Conv2D(filters=16, kernel_size=(3, 3), activation='relu', padding='same'))
model.add(MaxPooling2D(pool_size=(2, 2)))
model.add(Flatten())

model.add(Dense(units=32, activation='relu', kernel_regularizer=regularizers.l2(0.01)))
model.add(Dense(units=16, activation='relu', kernel_regularizer=regularizers.l2(0.01)))

model.add(Dense(units=2, activation='softmax'))
```

```python
model.compile(optimizer=Adam(learning_rate=0.0001), loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
model.fit(X_train, y_train, validation_data=(X_test, y_test), epochs=10)
```

```python
plt.plot(model.history.history['loss'])
plt.plot(model.history.history['val_loss'])
```

```python
plt.plot(model.history.history['accuracy'])
plt.plot(model.history.history['val_accuracy'])
```
